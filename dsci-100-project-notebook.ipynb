#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
wins_summary <- tidy_data %>%
summarize(w_mean = mean(conf_wins),
         w_min = min(conf_wins),
          w_max = max(conf_wins))


pct_summary <- tidy_data %>%
summarize(pct_mean = mean(conf_wins_pct),
         pct_min = min(conf_wins_pct),
          pct_max = max(conf_wins_pct))

total_summary<- tidy_data %>%
summarize(total_mean = mean(total_wins_pct),
         total_min = min(total_wins_pct),
          total_max = max(total_wins_pct))

summary_all <- bind_rows(wins_summary,pct_summary,total_summary) %>%
mutate(variable = c("conf_wins","conf_wins_pct","total_wins_pct")) %>%
select(variable, w_mean:total_max) %>%
rowwise()%>%
mutate(mean = sum(w_mean,pct_mean,total_mean,na.rm=TRUE),
       min = sum(w_min,pct_min,total_min,na.rm=TRUE),
       max = sum(w_max,pct_max,total_max,na.rm=TRUE)) %>%
select(variable,mean,min,max)

summary_all
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
number_miss <- data %>%
summarize(year = sum(is.na(year)),
          school= sum(is.na(school)),
          seed= sum(is.na(seed)),
          conference= sum(is.na(conference)),
          conf_wins= sum(is.na(conf_wins)),
          conf_losses= sum(is.na(conf_losses)),
          conf_wins_pct=sum(is.na(conf_wins_pct)),
          conf_rank=sum(is.na(conf_rank)),
          division=sum(is.na(division)),
          reg_wins=sum(is.na(reg_wins)),
          reg_losses=sum(is.na(reg_losses)),
          reg_wins_pct=sum(is.na(reg_wins_pct)),
          bid=sum(is.na(bid)),
          first_game_at_home=sum(is.na(first_game_at_home)),
          tourney_wins=sum(is.na(tourney_wins)),
          tourney_losses=sum(is.na(tourney_losses)),
          tourney_finish=sum(is.na(tourney_finish)),
          total_wins=sum(is.na(total_wins)),
          total_losses=sum(is.na(total_losses)),
          total_wins_pct=sum(is.na(total_wins_pct))) %>%
pivot_longer(year:total_wins_pct, 
             names_to = "variable",
             values_to = "miss")


number_miss
```

```r
numeric_plot <- tidy_data %>%
ggplot(aes(x=conf_wins_pct))+
geom_histogram(binwidth = 5, color = "black")+
labs(
    title = "Distribution of Conference Win Percentage",
    x = "Conference Win Percentage (%)",
    y = "Count")+
theme(text = element_text(size = 17))

numeric_plot
```

```r
options(repr.plot.width = 8, repr.plot.height = 8)
bid_rel_pro <- tidy_data %>%
group_by(bid)%>%
summarize(n = n()) %>%
mutate(proportion = n/sum(n))

categorical_plot <- bid_rel_pro %>%
ggplot(aes(x = bid, y=proportion, fill = bid)) +
geom_bar(stat = "identity")+
labs(
    title = "Relative proportions of different Bid",
    x = "Bid Type",
    y = "Proportion",
    fill = "Bid Type")+
theme(text = element_text(size = 17))
categorical_plot 
```

In this visualization, we compare the number of schools' bid types across different conference winning percentages. Schools with higher conference winning percentages (>= 90%) are more likely to get auto bids; schools with lower conference winning percentages are more likely to get at-large bids. This information may be useful for predicting the school's bid type based on its conference win percentage.

```r
options(repr.plot.width = 8, repr.plot.height = 8)

win_pct_plot <- tidy_data %>%
ggplot(aes(x = conf_wins_pct, fill = bid))+
geom_histogram(binwidth = 15, position = "dodge")+
labs(
    title = "Distribution of Conference Win Percentage of different Bid Type",
    x = "Conference Win Percentage (%)",
    y = "Count",
    fill = "Bid Type")+
theme(text = element_text(size = 17))
win_pct_plot 
```

In this visualization, we compare the average total win percentage of different bid types. Overall, schools that get auto bid have a higher average total win percentage than the average total win percentage of schools that get an at-large bid. This information may be useful for predicting the school's bid type based on its total win percentage.

```r
options(repr.plot.width = 8, repr.plot.height = 8)

total_mean_plot <- tidy_data %>%
group_by(bid) %>%
summarize(mean_total = mean(total_wins_pct)) %>%
ggplot(aes(x = bid, y = mean_total, fill = bid))+
geom_bar(stat = "identity")+
labs(
    title = "Average Total Win Percentage of different Bid Type",
    x = "Bid Type",
    y = "Average Total Win Percentage (%)",
    fill = "Bid Type")+
theme(text = element_text(size = 15))
total_mean_plot
```
